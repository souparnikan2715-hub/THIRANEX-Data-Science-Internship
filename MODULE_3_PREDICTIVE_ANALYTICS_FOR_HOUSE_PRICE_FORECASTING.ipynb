```python
# ============================================================
# MODULE 3: PREDICTIVE ANALYTICS FOR HOUSE PRICE FORECASTING
# THIRANEX DATA SCIENCE INTERNSHIP
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ------------------------------------------------------------
# 1. CREATE SAMPLE HOUSE PRICE DATASET
# ------------------------------------------------------------

np.random.seed(42)

n = 100

house_df = pd.DataFrame({
    "Area_sqft": np.random.randint(600, 3500, n),
    "Bedrooms": np.random.randint(1, 6, n),
    "Bathrooms": np.random.randint(1, 4, n),
    "Age_years": np.random.randint(1, 30, n)
})

# Create realistic house prices based on the features
noise = np.random.normal(0, 50000, n)

house_df["Price"] = (
    house_df["Area_sqft"] * 300
    + house_df["Bedrooms"] * 100000
    + house_df["Bathrooms"] * 80000
    - house_df["Age_years"] * 20000
    + 200000
    + noise
)

house_df["Price"] = house_df["Price"].round(2)

print("=" * 60)
print("HOUSE PRICE FORECASTING USING LINEAR REGRESSION")
print("=" * 60)

print("\nDataset Shape:", house_df.shape)

print("\nFirst 5 Records:")
display(house_df.head())

# ------------------------------------------------------------
# 2. DATASET INFORMATION
# ------------------------------------------------------------

print("\nMissing Values:")
print(house_df.isnull().sum())

print("\nDataset Statistics:")
display(house_df.describe())

# ------------------------------------------------------------
# 3. SAVE HOUSE PRICE DATASET
# ------------------------------------------------------------

house_df.to_csv(
    "house_price_dataset.csv",
    index=False
)

print("\nhouse_price_dataset.csv created successfully!")

# ------------------------------------------------------------
# 4. SELECT FEATURES AND TARGET
# ------------------------------------------------------------

X = house_df[
    [
        "Area_sqft",
        "Bedrooms",
        "Bathrooms",
        "Age_years"
    ]
]

y = house_df["Price"]

print("\nInput Features:")
print([
    "Area_sqft",
    "Bedrooms",
    "Bathrooms",
    "Age_years"
])

print("\nTarget Variable:")
print("Price")

# ------------------------------------------------------------
# 5. TRAIN-TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\n" + "=" * 60)
print("TRAIN-TEST SPLIT")
print("=" * 60)

print("Training Records:", len(X_train))
print("Testing Records :", len(X_test))

# ------------------------------------------------------------
# 6. TRAIN LINEAR REGRESSION MODEL
# ------------------------------------------------------------

model = LinearRegression()

model.fit(
    X_train,
    y_train
)

print("\nLinear Regression model trained successfully!")

# ------------------------------------------------------------
# 7. MAKE PREDICTIONS
# ------------------------------------------------------------

y_pred = model.predict(X_test)

print("\nPredictions generated successfully!")

# ------------------------------------------------------------
# 8. ACTUAL VS PREDICTED RESULTS
# ------------------------------------------------------------

prediction_results = pd.DataFrame({
    "Actual_Price": y_test.values,
    "Predicted_Price": y_pred.round(2)
})

print("\n" + "=" * 60)
print("ACTUAL VS PREDICTED HOUSE PRICES")
print("=" * 60)

display(prediction_results.head(10))

# ------------------------------------------------------------
# 9. MODEL EVALUATION
# ------------------------------------------------------------

mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

r2 = r2_score(
    y_test,
    y_pred
)

print("\n" + "=" * 60)
print("MODEL EVALUATION")
print("=" * 60)

print(f"Mean Absolute Error  : {mae:.2f}")
print(f"Root Mean Squared Error : {rmse:.2f}")
print(f"R² Score             : {r2:.4f}")

# ------------------------------------------------------------
# 10. SAVE PREDICTION RESULTS
# ------------------------------------------------------------

prediction_results.to_csv(
    "house_price_predictions.csv",
    index=False
)

print("\nhouse_price_predictions.csv created successfully!")

# ------------------------------------------------------------
# 11. ACTUAL VS PREDICTED PRICE GRAPH
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    s=70
)

# Reference diagonal line
minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.title("Actual vs Predicted House Prices")
plt.xlabel("Actual House Price")
plt.ylabel("Predicted House Price")

plt.grid(True)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 12. FEATURE COEFFICIENTS
# ------------------------------------------------------------

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

print("\n" + "=" * 60)
print("MODEL FEATURE COEFFICIENTS")
print("=" * 60)

display(coefficients)

# ------------------------------------------------------------
# 13. SAMPLE HOUSE PRICE PREDICTION
# ------------------------------------------------------------

sample_house = pd.DataFrame({
    "Area_sqft": [2000],
    "Bedrooms": [3],
    "Bathrooms": [2],
    "Age_years": [5]
})

sample_prediction = model.predict(sample_house)

print("\n" + "=" * 60)
print("SAMPLE HOUSE PRICE PREDICTION")
print("=" * 60)

print("Area       : 2000 sqft")
print("Bedrooms   : 3")
print("Bathrooms  : 2")
print("Age        : 5 years")
print(f"Predicted Price: {sample_prediction[0]:.2f}")

# ------------------------------------------------------------
# 14. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MODULE 3 COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nGenerated Files:")
print("✓ house_price_dataset.csv")
print("✓ house_price_predictions.csv")

print("\nModel: Linear Regression")
print(f"R² Score: {r2:.4f}")
```
